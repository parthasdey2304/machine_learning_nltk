```python
# importing nltk library and its punkt_tab
import nltk
nltk.download('punkt_tab')
```

```python
# making tokenization of words for these.
# converting sentence full of words into tokens
from nltk.tokenize import word_tokenize
text = 'I love learning NLTK'
tokens = word_tokenize(text)
print(tokens)
```

```python
# using sent_tokenize to make a list of all sentences in thee text
from nltk.tokenize import sent_tokenize
text = 'I am learning Machine Learning! Using the Python programming language.'
sent_tokens = sent_tokenize(text)
print(sent_tokens)
```

```python
# using RegexpTokenizer to get the words using Regex
from nltk.tokenize import RegexpTokenizer
text = 'I love leaning NLTK! It\'s very interesting'
tokenses = RegexpTokenizer(r'\w+')
token = tokenses.tokenize(text)
print(token)
```
